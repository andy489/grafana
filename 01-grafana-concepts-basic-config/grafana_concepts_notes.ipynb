## Module 2 — Installing and Configuring Grafana

### Grafana Cloud

> *"A free plan that's actually useful."*

- Free tier available at https://grafana.com/products/cloud
- Quickest way to get started without managing infrastructure

### Linux Installation

#### Key Resources

| Resource | URL |
|----------|-----|
| Installation docs (Debian) | https://grafana.com/docs/grafana/latest/setup-grafana/installation/debian/ |
| Start/restart docs | https://grafana.com/docs/grafana/latest/setup-grafana/start-restart-grafana/ |
| Default login page | `http://<hostname-or-ip>:3000/login` |

```python
# Example: install Grafana on Debian/Ubuntu
# (run these in a terminal, not in this notebook)

install_steps = """
sudo apt-get install -y apt-transport-https software-properties-common wget
sudo mkdir -p /etc/apt/keyrings/
wget -q -O - https://apt.grafana.com/gpg.key | gpg --dearmor | sudo tee /etc/apt/keyrings/grafana.gpg > /dev/null
echo "deb [signed-by=/etc/apt/keyrings/grafana.gpg] https://apt.grafana.com stable main" | sudo tee -a /etc/apt/sources.list.d/grafana.list
sudo apt-get update
sudo apt-get install grafana
sudo systemctl start grafana-server
sudo systemctl enable grafana-server
"""
print(install_steps)
```

### Other Installation Options

- **Windows** — MSI installer

## Module 3 — Adding a Data Source to Grafana

### What Is a Data Source?

> *Where Grafana pulls data from to create dashboards and visualizations.*

Reference: https://grafana.com/docs/grafana/latest/datasources

### Aggregating Multiple Data Sources

- A single dashboard can combine **more than one type of data**
- Mix **infrastructure and business metrics**
- Use data sources across **single or multiple dashboard panels**

### What Is Prometheus?

| Attribute | Detail |
|-----------|--------|
| **Type** | Open-source monitoring and alerting toolkit |
| **Storage** | Metrics as time series data |
| **Collection** | Scrapes data at regular intervals |
| **Query language** | PromQL |

### Prometheus Data Source Setup

| Endpoint | URL Pattern |
|----------|-------------|
| Node Exporter metrics | `http://<IP or Hostname>:9100/metrics` |
| Prometheus targets | `http://<IP or Hostname>:9090/targets` |
| Node Exporter install docs | https://prometheus.io/docs/guides/node-exporter/#installing-and-running-the-node-exporter |

```python
import json

# Example Grafana API payload to add a Prometheus data source
datasource_payload = {
    "name": "Prometheus",
    "type": "prometheus",
    "url": "http://localhost:9090",
    "access": "proxy",
    "isDefault": True
}

print("POST /api/datasources")
print(json.dumps(datasource_payload, indent=2))
```

### Pre-built Dashboards

- Browse the dashboard library: https://grafana.com/grafana/dashboards
- Dashboards are pre-configured for specific data sources
- **Import as-is** or modify to fit your needs — saves significant time
